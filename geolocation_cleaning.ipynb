```python
#Check jumlah missing value di setiap kolom  
print('geolocation_zip_code_prefix: ', dg['geolocation_zip_code_prefix'].isna().sum())
print('geolocation_lat: ', dg['geolocation_lat'].isna().sum())
print('geolocation_lng: ', dg['geolocation_lng'].isna().sum())
print('geolocation_city: ', dg['geolocation_city'].isna().sum()) 
print('geolocation_state: ', dg['geolocation_state'].isna().sum()) 
```

```python
dg['geolocation_state'].value_counts(dropna = False) #Check kategori penulisan kolom geolocation_state  
```
