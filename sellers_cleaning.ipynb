```python
#Check jumlah missing value di setiap kolom  
print('seller_id: ', dc['seller_id'].isna().sum())
print('seller_zip_code_prefix: ', dc['seller_zip_code_prefix'].isna().sum())
print('seller_city: ', dc['seller_city'].isna().sum())
print('seller_state: ', dc['seller_state'].isna().sum())
```

```python
state_mapping = {
    'AC': 'Acre',
    'AL': 'Alagoas',
    'AM': 'Amazonas',
    'AP': 'Aapá',
    'BA': 'Bahia',
    'CE': 'Ceará',
    'DF': 'Distrito Federal',
    'ES': 'Espírito Santo',
    'GO': 'Goiás',
    'MA': 'Maranhão',
    'MG': 'Minas Gerais',
    'MS': 'Mato Grosso do Sul',
    'MT': 'Mato Grosso',
    'PA': 'Pará',
    'PB': 'Paraíba',
    'PE': 'Pernambuco',
    'PI': 'Piauí',
    'PR': 'Paraná',
    'RJ': 'Rio de Janeiro',
    'RN': 'Rio Grande do Norte',
    'RO': 'Rondônia',
    'RR': 'Roraima',
    'RS': 'Rio Grande do Sul',
    'SC': 'Santa Catarina',
    'SE': 'Sergipe',
    'SP': 'São Paulo',
    'TO': 'Tocantins'
}
dc['seller_state']=dc['seller_state'].map(state_mapping) 
dc['seller_state'].value_counts(dropna=False) 
```
