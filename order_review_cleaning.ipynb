```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dc = pd.read_csv('raw_dataset/olist_order_reviews_dataset.csv') 
dc.info() 
dc.head()
```

```python
dc.columns.is_unique #Check data duplikat di setiap kolom (output: True) 
```

```python
#Check jumlah missing value di setiap kolom  
print('order_id: ', dc['order_id'].isna().sum())
print('review_id: ', dc['review_id'].isna().sum())
print('review_score: ', dc['review_score'].isna().sum())
print('review_comment_title: ', dc['review_comment_title'].isna().sum()) 
print('review_comment_message: ', dc['review_comment_message'].isna().sum()) 
print('review_creation_date: ', dc['review_creation_date'].isna().sum()) 
print('review_answer_timestamp: ', dc['review_answer_timestamp'].isna().sum()) 
```

```python
def comment_title (x):
    if pd.isna(x):
        return 'No Comment Title'

    x = str(x).lower() 
    if 'super' in x and 'recomendo' in x :
        return 'Super recomendo'
    elif 'recomendo' in x:
        return 'Recomendo' 

    return x.title().strip() 
    
dc['review_comment_title'] = dc['review_comment_title'].apply(comment_title) 
dc['review_comment_title'].value_counts(dropna=False) 
```

```python
#dc['review_comment_message']=dc['review_comment_message'].fillna('No Comment Message') 
def comment_message (x):
    if pd.isna(x):
        return 'No Comment Message'

    x = str(x).lower() 
    if 'muito' in x and 'bom' in x or 'otimo' in x: 
        return 'Muito bom' 
    elif 'bom' in x:
        return 'Bom'  
    elif 'super' in x and 'recomendo' in x :
        return 'Super recomendo'
    elif 'recomendo' in x:
        return 'Recomendo' 

    return x.title().strip() 

dc['review_comment_message']=dc['review_comment_message'].apply(comment_message) 
dc['review_comment_message'].value_counts(dropna=False)
```
