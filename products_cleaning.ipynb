```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dc = pd.read_csv('raw_dataset/olist_products_dataset.csv') 
dc.info() 
dc.head()
```

```python
dc.columns.is_unique #Check data duplikat di setiap kolom (output: True) 
```

```python
#Check jumlah missing value di setiap kolom  
print('product_id: ', dc['product_id'].isna().sum())
print('product_category_name: ', dc['product_category_name'].isna().sum())
print('product_name_lenght: ', dc['product_name_lenght'].isna().sum())
print('product_description_lenght: ', dc['product_description_lenght'].isna().sum())
print('product_photos_qty: ', dc['product_photos_qty'].isna().sum())
print('product_weight_g: ', dc['product_weight_g'].isna().sum())
print('product_length_cm: ', dc['product_length_cm'].isna().sum())
print('product_height_cm: ', dc['product_height_cm'].isna().sum())
print('product_width_cm: ', dc['product_width_cm'].isna().sum())
```

```python
dc['product_category_name']=dc['product_category_name'].fillna('Unknown Category Name') 
dc['product_category_name'].value_counts(dropna=False) 
```

```python
dc['product_name_lenght']=dc['product_name_lenght'].fillna(0) 
dc['product_name_lenght'].value_counts(dropna=False) 
```

```python
dc['product_description_lenght']=dc['product_description_lenght'].fillna(0) 
dc['product_description_lenght'].value_counts(dropna=False) 
```

```python
dc['product_photos_qty']=dc['product_photos_qty'].fillna(0) 
dc['product_photos_qty'].value_counts(dropna=False) 
```

```python
dc['product_weight_g']=dc['product_weight_g'].fillna(dc['product_weight_g'].median())
dc['product_weight_g'].value_counts(dropna=False) 
```

```python
dc['product_length_cm']=dc['product_length_cm'].fillna(dc['product_length_cm'].median())
dc['product_length_cm'].value_counts(dropna=False) 
```

```python
dc['product_height_cm']=dc['product_height_cm'].fillna(dc['product_height_cm'].median())
dc['product_height_cm'].value_counts(dropna=False) 
```

```python
dc['product_width_cm']=dc['product_width_cm'].fillna(dc['product_width_cm'].median())
dc['product_width_cm'].value_counts(dropna=False) 
```

```python
dc.info()
dc.head()
```

```python
dc['product_name_lenght']=dc['product_name_lenght'].astype(int)
dc['product_description_lenght']=dc['product_description_lenght'].astype(int)
dc['product_photos_qty']=dc['product_photos_qty'].astype(int)
dc.info()
```

```python
dc['product_category_name']=dc['product_category_name'].str.replace('_',' ') 
dc['product_category_name'].value_counts(dropna=False) 
```
